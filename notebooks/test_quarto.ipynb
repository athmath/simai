# The EOQ Model

Suppose annual demand is $$D$, ordering cost is $K$,
and holding cost is $h$.

The total relevant cost is

$$
C(Q)=\frac{KD}{Q}+\frac{hQ}{2}.
$$

Differentiating gives

$$
C'(Q)=-\frac{KD}{Q^2}+\frac{h}{2},
$$

and therefore

$$
Q^*=\sqrt{\frac{2KD}{h}}.
$$

```python
import numpy as np

D = 1000
K = 50
h = 2

Q_star = np.sqrt(2 * K * D / h)
Q_star
```

```python
import matplotlib.pyplot as plt

Q = np.linspace(20, 500, 500)
cost = K * D / Q + h * Q / 2

plt.plot(Q, cost)
plt.xlabel("Order quantity Q")
plt.ylabel("Annual cost")
plt.show()
```
