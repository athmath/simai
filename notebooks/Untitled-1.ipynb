```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def line_equation_1(m, x0, y0):   
    """Returns the coefficients (A, B) of the line equation y=ax+b
    given the slope m and a point (x0, y0) on the line."""      
    A = m
    B = y0 - m * x0
    return A, B

def line_equation_2(x1, y1, x2, y2):
    """Returns the coefficients (A, B) of the line equation y=ax+b
    given two points (x1, y1) and (x2, y2) on the line."""      
    if x2 == x1:
        raise ValueError("x1 and x2 cannot be the same value.")
    m = (y2 - y1) / (x2 - x1)
    A = m
    B = y1 - m * x1
    return A, B




def g(x):
    return (x**2 - x - 6)**2
def g1(x):
    return x-3

```

```python
print(line_equation_2(0,1/2,-3,5))  # Example usage
```

```python
g(3.039)
```

```python
x_values = np.linspace(2, 4, 100)
y1_values = g(x_values)
y2_values = g1(x_values)    
plt.plot(x_values, y1_values, label='g(x) = (x^2 - x - 6)^2')
plt.plot(x_values, y2_values, label='g1(x) = x - 3')
plt.axhline(0, color='black',linewidth=0.5, ls='--')
plt.axvline(0, color='black',linewidth=0.5, ls='--')
plt.title('Graphs of g(x) and g1(x)')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid()
plt.show()
```
