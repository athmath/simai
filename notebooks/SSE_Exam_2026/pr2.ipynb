```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def smoking_probability(age):
    return 0.18+(age-15)*(59-age)**2/1e6
```

```python
x=np.arange(10, 80)
y=smoking_probability(x)
plt.plot(x, y)
plt.xlabel('Age')
plt.ylabel('Probability of Smoking')
plt.title('Probability of Smoking vs Age')
plt.grid()
plt.show()
```

```python
def smoking_simulation(lbage, ubage, num_people):
    ages = np.random.uniform(lbage, ubage, num_people)
    smokers = np.random.rand(num_people) < smoking_probability(ages)
    probability_smokers = np.mean(smokers)
    young_smokers = (smokers==True) & (ages >16) & (ages < 30)
    prob_young_smokers = np.sum(young_smokers)/np.sum(smokers)
    return probability_smokers, prob_young_smokers
```

```python
lbage = 16
ubage = 59
num_people = 100000
prob_smokers, prob_young_smokers = smoking_simulation(lbage, ubage, num_people)
print(f"Estimated Probability of Smoking: {prob_smokers:.4f}")
print(f"Estimated Probability of Young Smokers (16-30): {prob_young_smokers:.4f}")
```
