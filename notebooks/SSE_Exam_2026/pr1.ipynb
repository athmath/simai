```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def game_simulation(p, throws, cost, reward_per_win, threshold, reward_above_threshold):
    wins = 0
    for _ in range(throws):
        if np.random.rand() < p:
            wins += 1
    total_reward = wins * reward_per_win
    if wins > threshold:
        total_reward += reward_above_threshold
    return total_reward - cost
```

```python
game_simulation(p=0.5, throws=10, cost=5, reward_per_win=2, threshold=6, reward_above_threshold=10) 
```

```python
n=5000
results = [game_simulation(p=0.7, throws=20, cost=24, reward_per_win=2, threshold=15, reward_above_threshold=200) for _ in range(n)]
#plt.hist(results, bins=20, edgecolor='black')
mean_reward = np.mean(results)
print(f"Average reward over {n} simulations: {mean_reward:.2f}")
loss = [1 if r < 0 else 0 for r in results]
prob_loss = np.mean(loss)
print(f"Probability of losing money: {prob_loss:.2%}")
stdev_reward = np.std(results)
lower_bound = mean_reward - 1.96 * stdev_reward / np.sqrt(n)
upper_bound = mean_reward + 1.96 * stdev_reward / np.sqrt(n)
print(f"95% confidence interval for the average reward: [{lower_bound:.2f}, {upper_bound:.2f}]")
stdev_loss = np.sqrt(prob_loss * (1 - prob_loss) / n)
lower_bound_loss = prob_loss - 1.96 * stdev_loss
upper_bound_loss = prob_loss + 1.96 * stdev_loss
print(f"95% confidence interval for the probability of losing money: [{lower_bound_loss:.2%}, {upper_bound_loss:.2%}]")
```
